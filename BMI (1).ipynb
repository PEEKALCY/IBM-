```python
# BMI Calculator in Jupyter Notebook (No Tkinter)

# Pre-set Access Code
ACCESS_CODE = "12345"

# Function to calculate BMI
def calculate_bmi():
    try:
        # Ask for access code
        code = input("Enter Access Code: ")
        if code != ACCESS_CODE:
            print("❌ Access Denied! Invalid Code.")
            return

        # Get weight & height
        weight = float(input("Enter your weight (lbs): "))
        feet = float(input("Enter your height - feet: "))
        inches_input = input("Enter your height - inches (optional): ")

        # Default inches to 0 if blank
        inches = float(inches_input) if inches_input.strip() != "" else 0

        # Convert height to total inches
        height_in_inches = (feet * 12) + inches
        if height_in_inches <= 0 or weight <= 0:
            print("❌ Error: Height and Weight must be greater than 0.")
            return

        # Convert to metric
        height_m = height_in_inches * 0.0254
        weight_kg = weight * 0.453592

        # Calculate BMI
        bmi = weight_kg / (height_m ** 2)

        # Classification
        if bmi < 18.5:
            classification = "Underweight"
            risk = "Health risk: Increased"
        elif bmi < 25:
            classification = "Normal weight"
            risk = "Health risk: Normal"
        elif bmi < 30:
            classification = "Overweight"
            risk = "Health risk: Increased"
        elif bmi < 35:
            classification = "Obese (Class I)"
            risk = "Health risk: High"
        elif bmi < 40:
            classification = "Obese (Class II)"
            risk = "Health risk: Very High"
        else:
            classification = "Obese (Class III)"
            risk = "Health risk: Extremely High"

        # Print results
        print("\n✅ BMI Calculation Complete")
        print(f"Your BMI: {bmi:.2f}")
        print(f"Classification: {classification}")
        print(f"{risk}")

    except ValueError:
        print("❌ Input Error: Please enter numbers only.")

# Run the calculator
calculate_bmi()


```
